```python
import praw
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
reddit = praw.Reddit(
    client_id = os.getenv('REDDIT_CLIENT_I'),
    clinet_secrte = os.getenv('REDDIT_CLIENT_SECRET'),
    user_agent = os.getenv('REDDIT_USER_AGENT')
)
```

```python
f1_sub = reddit.subreddit('formula1')
print(f"Subreddit: {f1_sub.display_name}")
print(f"Subscribers: {f1_sub.subscribers:,}")
```

```python
for post in f1_sub.hot(limit=1):
    print(f"Top post: {post.title}")
```
